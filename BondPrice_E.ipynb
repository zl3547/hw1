```python
def getBondPrice_E(face, couponRate, yc):
    bondPrice_E = 0
    pvCoupons = 0
    cf = couponRate*face
    for i, x in enumerate(yc):
        pvCoupons+=cf*(1+x)**-(i+1)
    bondPrice_E = pvCoupons+face*(1+yc[len(yc)-1])**-len(yc)
    return bondPrice_E

# Test values
yc = [.010,.015,.020,.025,.030]
face = 2000000
couponRate = .04

bondPrice_E = getBondPrice_E(face, couponRate, yc)
print(bondPrice_E)
```
