```python
def getBondPrice_Z(face, couponRate, times, yc):
    BondPrice = 0
    pvCoupons = 0
    cf = couponRate * face
    for (i,x) in zip(times,yc):
        pvCoupons += cf * (1 + x) ** - i
    BondPrice = pvCoupons + face * (1 + yc[len(yc)-1]) ** - times[len(yc)-1]
    return BondPrice

# Test values
yc = [.010,.015,.020,.025,.030]
times=[1,1.5,3,4,7]
face = 2000000
couponRate = .04

bondPrice_Z = getBondPrice_Z(face, couponRate, times, yc)
print(bondPrice_Z)
```
